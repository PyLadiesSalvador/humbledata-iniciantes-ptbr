## Sobre o Seaborn
Seaborn é uma biblioteca Python de visualização de dados baseada no matplotlib. Ela fornece uma interface de alto nível para desenho de gráficos estatísticos informativos e atrativos, sendo poderosa para visualização de dados categóricos.

![](https://d1rwhvwstyk9gu.cloudfront.net/2017/07/seaburn-1.png)

Vamos usar o [Pokemon.csv](https://gist.github.com/armgilles/194bcff35001e7eb53a2a8b441e8b2c6). Vamos dar uma olhada nos dados:

```python
import pandas as pd

pokemon_df = pd.read_csv('data/Pokemon/pokemon.csv', index_col=0)
pokemon_df.head(10)
```

Por exemplo, queremos comparar o Ataque de diferentes tipos de Pokemon, para ver se algum tipo é mais poderoso do que outros no geral:

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.catplot(x="Type 1", y="Attack", data=pokemon_df);
```

Na importação geralmente abreviamos 'seaborn' como 'sns'. (É uma referência a [West Wing / Rob Lowe](https://en.wikipedia.org/wiki/Sam_Seaborn)!) Note que também temos que importar matplotlib.pyplot porque Seaborn é uma biblioteca construída sobre o matplotlib. Nós temos um gráfico mas ele está feio e ilegível, vamos adicionar algumas configurações para deixá-lo mais agradável

**Tente adicionar `aspect=2.5` como último argumento na seguinte função `sns.catplot`**

```python
sns.catplot(x="Type 1", y="Attack", data=pokemon_df);
```

Então você pode ver que adicionando 'aspect' tornamos o gráfico mais largo. A largura do gráfico é igual a 'aspect * height', então adicionando 'aspect' aumentamos a largura do gráfico. Essa é uma das configurações que podemos adicionar no gráfico. Para a lista completa com detalhes, podemos seguir a [documentação oficial](https://seaborn.pydata.org/generated/seaborn.catplot.html#seaborn.catplot), mas vamos fazer uma introdução às configurações mais comuns.

Por exemplo, aqui podemos ver um deslocamento aleatório no eixo x para todos os pontos, sem que os marcadores se sobreponham uns sobre os outros. Isso é feito pela configuração do 'jitter', que por padrão é True (verdadeiro). Vamos desativá-la e ver como fica:

**Tente adicionar `jitter=False` como último argumento na seguinte função `sns.catplot`**

```python
sns.catplot(x="Type 1", y="Attack", data=pokemon_df, aspect=2.5);
```

Então agora temos um gráfico em que os pontos estão alinhados de acordo com suas categorias sem o deslocamento no eixo x. A decisão sobre qual ponto utilizar depende do quão importante é visualizar a distribuição dos valores (exemplo: Ataque). No nosso caso, queremos saber como o valor de ataque (Attack) é distribuido para cada tipo de pokemon (Type), então talvez seja útil ter o 'jitter' ativado, ou melhor ainda, espalhar os pontos ainda mais para mostrar a distribuição de melhor forma:

**Tente adicionar `kind="swarm"` como último argumento na seguinte função `sns.catplot`**

```python
sns.catplot(x="Type 1", y="Attack", data=pokemon_df, aspect=2.5);
```

Aqui, podemos fazer isso definindo kind='swarm', para que os pontos não fiquem sobrepostos. A desvantagem é que esse tipo de gráfico exige mais espaço horizontal. Imagine que não queremos deixá-lo muito largo por causa das limitações de espaço no documento. Nesse caso, podemos girar o gráfico em 90 graus invertendo os eixos x e y — e então ajustar o aspecto (aspect) e a altura (height) conforme necessário.

**Tente trocar `x` e `y`, e adicionar `height=12, aspect=0.6, kind="swarm"` nos argumentos da seguinte função `sns.catplot`**


```python
sns.catplot(x="Type 1", y="Attack", data=pokemon_df);
```

Até aqui, temos algumas coisas para observar:

1. Para alguns tipos, como Psychic, existe um intervalo de ataque (Attack) muito amplo, com uma “cauda” longa no final (alguns pokemons do tipo Physic possuem poder de ataque muilo alta, enquanto a maioria não)

2. Por outro lado, o tipo Poison ficam majoritariamente na faixa de 40 a 110 de ataque.

3. No geral, Pokémon do tipo Dragon têm mais poder de ataque do que os do tipo Fairy, mas existem dois Pokémon Fairy que possuem maior ataque.

Entretanto, gostaríamos de ir mais fundo: eu tenho uma teoria que os pokemons lendários (Legendary Pokemon) são mais poderosos. Vamos colorir o gráfico de acordo com a tag 'Legendary' e ver se existe alguma relação ou não entre o o pokemon ser do tipo Legendary e seu ataque (Attack):

**Tente adicionar`hue="Legendary"` como último argumento na seguinte função `sns.catplot`**

```python
plt.figure(figsize=(15, 6))
sns.stripplot(x="Type 1", y="Attack", data=pokemon_df, size=7)
```

Ah ha! Vemos que muitos do tipo Psychic que possuem um valor de ataque (Attack) mais alto que outros são na verdade pokemons lendários. Isso também acontece com os tipos Flying e Ground.

Às vezes queremos ter múltiplos gráficos plotados juntos em uma mesma visualização para facilitar a comparação. Um jeito de fazer isso no seaborn é usando FacetGrid. A classe FacetGrid é útil quando queremos visualizar a distribuição de uma variável, ou a relação entre  múltiplas variáveis,  separadamente em subconjuntos do seu conjunto de dados principal. A seguir, vamos usar FacetGrid para verificar se existe alguma diferença, em relação à análise feita acima, quando observamos diferentes Generations (gerações de pokemon).

Para criar uma FacetGrid, podemos fazer o seguinte:

```python
g = sns.FacetGrid(pokemon_df, col="Generation")
```

Aqui temos 6 áreas de plotagem que equivalem ao número de diferentes Generations que temos (podemos checar quais são as diferentes gerações assim):

```python
pokemon_df["Generation"].unique()
```

Contudo, gostaríamos de ter os gráficos alinhados verticalmente em vez de horizontalmente.

**Tente substituir `col` por `row`na seguinte instância `sns.FacetGrid`**

```python
g = sns.FacetGrid(pokemon_df, col="Generation")
```

Ok, agora temos o layout, mas como vamos colocar os gráficos aqui? Para alguns gráficos, isso pode ser fieto com o método [FacetGrid.map()](https://seaborn.pydata.org/generated/seaborn.FacetGrid.map.html#seaborn.FacetGrid.map), por exemplo, usando sns.countplot para contar quantos pokemons existem em diferentes espécies.

```python
g = sns.FacetGrid(pokemon_df, row="Generation", aspect=3.5)
g.map(sns.countplot, "Type 1");
```

Mas com sns.catplot que usamos antes, isso fica ainda mais simples. Como catplot já é um objeto FacetGrid, podemos diretamente adicionar o atributo `row` ou `col` a ele.

**Tente adicionar `row="Generation"` como último argumento na seguinte função `sns.catplot`**

```python
plt.figure(figsize=(15, 6))
sns.stripplot(x="Type 1", y="Attack", data=pokemon_df, size=7, hue="Legendary")
```

Agora você pode ver que em cada Generation, os pokemons lendários são outliers com super poderes de ataque comparados com outros dentro da sua mesma geração. Para mais detalhes usando FacetGrids, você pode checar a documentação oficial aqui: https://seaborn.pydata.org/tutorial/axis_grids.html

<div class="alert alert-block alert-warning" style="padding: 0px; padding-left: 20px; padding-top: 5px;"><h2 style="color: #301E40">
Pacotes de Análise de Dados
</h2><br>
</div>

Cientistas de dados usam uma grande variedade de bibliotecas Python que tornam o trabalho com dados significativamente mais fácil. Essas bibliotecas consistem principalmente em:

| Pacote | Descrição                                                                                                                                                                                                      |
| -- |----------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------|
| `NumPy` | Cálculos numéricos - faz todo o trabalho pesado passando para sub-rotinas em C. Isso significa que você obtém _tanto_ a produtividade do Python _quanto_ o poder computacional do C. O melhor dos dois mundos! |
| `SciPy` | Computação científica, testes estatísticos e muito mais!                                                                                                                                                       |
| `pandas` | Seu _pau-pra-toda-obra_ de manipulação de dados. Você provavelmente verá pandas usado em qualquer demonstração PyData! `pandas` é construído em cima do NumPy, então é bem **rápido**.                         |
| `matplotlib` | Um pacote antigo mas poderoso de visualização de dados, inspirado em Matlab.                                                                                                                                   |
| `Seaborn` | Um pacote mais novo e fácil de usar, mas limitado, de visualização de dados, construído em cima de `matplotlib`.                                                                                                 |
| `scikit-learn` | Sua loja única de aprendizado de máquina! Classificação, regressão, agrupamento, redução dimensional e muito mais.                                                                                             |
| `nltk` e `spacy` | nltk = kit de ferramentas de processamento de linguagem natural; spacy é um pacote mais novo para processamento de linguagem natural mas muito fácil de usar.                                                  |
| `statsmodels` | Testes estatísticos, previsão de séries temporais e muito mais. A interface "model formula" será familiar para usuários de R.                                                                                  |
| `requests` e `Beautiful Soup` | `requests` + `Beautiful Soup` = ótima combinação para criar web scrapers.                                                                                                                                      |
| `Jupyter` | Jupyter em si também é um pacote. Veja a versão mais recente em https://pypi.org/project/jupyter/, e atualize com `conda install jupyter==1.0.0`                                                               |

Embora existam inúmeros outros disponíveis, Por enquanto, focaremos principalmente na biblioteca que representa 99% do nosso trabalho: `pandas`.

Pandas é construído sobre a velocidade e poder do NumPy.

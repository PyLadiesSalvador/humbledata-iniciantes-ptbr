De acordo com a [documentação oficial](https://matplotlib.org/gallery/index.html):

`matplotlib.pyplot` é uma coleção de funções no estilo comando que fazem o Matplotlib funcionar como o MATLAB. Cada função do pyplot faz alguma alteração em uma figura: por exemplo, cria uma figura, cria uma área de plotagem em uma figura, plota linhas em uma área de plotagem, decora o gráfico com rótulos, etc.

O `pyplot` destina-se principalmente a plots interativos e casos simples de geração programática de gráficos.

![](https://miro.medium.com/max/2000/1*swPzVFGpYdijWAmbrydCDw.png)

### Usando DataFrame.plot() no pandas

O objeto DataFrame do pandas tem um método `plot()` que fornece gráficos básicos de vários tipos, incluindo: 'line', 'bar', 'hist', 'box' etc. Você também pode definir parâmetros para controlar o layout e os rótulos do gráfico.

O `plot()` usa `matplotlib.pyplot` em segundo plano, o que torna mais fácil plotar dados em um DataFrame.

Você encontrará esta página útil:
https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.DataFrame.plot.html


#### Exemplo: Gráfico de caixa (box plot) geral

```python
df.plot(kind='box')
```

As escalas dos nossos dados não se alinham muito bem. Então, para fins de plotagem, vamos ignorar a massa corporal `(body_mass_g)` dos pinguins.

```python
df.drop(["body_mass_g"], axis=1).plot(kind='box')
```

#### Melhor apresentação: tamanho da figura, título e legenda

```python
df.drop(["body_mass_g"], axis=1).plot(kind='box', figsize=(10,8), title='Box plot de diferentes medidas das espécies de pinguins', legend=True)
```

#### Fazendo subplots

```python
df.plot(kind='box',
        subplots=True, layout=(2,2),
        figsize=(10,8), title='Box plot de diferentes medidas das espécies de pinguins', legend=True)
```

### Separe os dados nas 3 espécies

```python
df['species'].unique()
```

```python
df_adelie = df[df['species'] == 'Adelie']
```

```python
df_chinstrap = df[df['species'] == 'Chinstrap']
```

```python
df_gentoo = df[df['species'] == 'Gentoo']
```

### Exemplo de gráfico de dispersão (scatter): comprimento `(bill_length_mm)` x profundidade do bico `(bill_depth_mm)`

```python
plt.scatter(df_adelie['bill_length_mm'], df_adelie['bill_depth_mm'], c='r')
plt.scatter(df_chinstrap['bill_length_mm'], df_chinstrap['bill_depth_mm'], c='g')
plt.scatter(df_gentoo['bill_length_mm'], df_gentoo['bill_depth_mm'], c='b')
```

#### Melhor apresentação: tamanho da figura, rótulos e legenda

```python
plt.figure(figsize=(10,8)) # define o tamanho do gráfico

plt.scatter(df_adelie['bill_length_mm'], df_adelie['bill_depth_mm'], c='r')
plt.scatter(df_chinstrap['bill_length_mm'], df_chinstrap['bill_depth_mm'], c='g')
plt.scatter(df_gentoo['bill_length_mm'], df_gentoo['bill_depth_mm'], c='b')

ax = plt.gca() # gca indica ao resto do código para referenciar o gráfico atual

ax.set_xlabel('Comprimento do Bico (mm)')
ax.set_ylabel('Profundidade do Bico (mm)')
ax.set_title('Comprimento e Profundidade do Bico por Espécie de Pinguim')

ax.legend(('adelie', 'chinstrap', 'gentoo'))
```

### Exemplo de histograma: Comprimento do bico

```python
plt.figure(figsize=(10,8))

plt.hist(df_adelie['bill_length_mm'], color='r', alpha=.5) # alpha define a transparência
plt.hist(df_chinstrap['bill_length_mm'], color='g', alpha=.5)
plt.hist(df_gentoo['bill_length_mm'], color='b', alpha=.5)

ax = plt.gca()

ax.set_xlabel('Comprimento do Bico (mm)')
ax.set_title('Histograma do Comprimento do Bico por Espécie de Pinguim')

ax.legend(('adelie', 'chinstrap', 'gentoo'))
```

### Exemplo de subplots

Fazer subplots usando apenas `plt` é um pouco mais complicado. É considerado mais avançado e requer entendimento dos blocos construtivos de um gráfico. Não se preocupe se achar desafiador — siga o exemplo e teste para entender melhor.

O exemplo abaixo plota histogramas de Comprimento do Bico e Profundidade do Bico lado a lado.


```python
# Primeiro, decidimos quantos subplots queremos e como orientá-los
# por exemplo, lado a lado (1 linha, 2 colunas)

fig, (ax0, ax1) = plt.subplots(nrows=1, ncols=2, figsize=(15,8))

# fig é o objeto figura (toda a área do gráfico)
# ax0 e ax1 são os eixos (subplots)

# Para ax0:
ax0.hist(df_adelie['bill_length_mm'], color='r', alpha=.5)
ax0.hist(df_chinstrap['bill_length_mm'], color='g', alpha=.5)
ax0.hist(df_gentoo['bill_length_mm'], color='b', alpha=.5)

ax0.set_xlabel('Comprimento do Bico (mm)')
ax0.set_title('Histograma do Comprimento do Bico por Espécie de Pinguim')

ax0.legend(('adelie', 'chinstrap', 'gentoo'))

# Para ax1:
ax1.hist(df_adelie['bill_depth_mm'], color='r', alpha=.5)
ax1.hist(df_chinstrap['bill_depth_mm'], color='g', alpha=.5)
ax1.hist(df_gentoo['bill_depth_mm'], color='b', alpha=.5)

ax1.set_xlabel('Profundidade do Bico (mm)')
ax1.set_title('Histograma da Profundidade do Bico por Espécie de Pinguim')

ax1.legend(('adelie', 'chinstrap', 'gentoo'))

plt.show() # depois de construir tudo, use show() para exibir os gráficos
```

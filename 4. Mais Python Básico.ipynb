Sets (Conjuntos)
</h2><br>
</div>

**Um conjunto é uma coleção de elementos únicos, não ordenados e não indexados.**

```python
conjunto_importante = set(['eu','meu eu', 'mim', 'eu'])
conjunto_importante
```

Nós podemos adicionar um item a um conjunto usando o método ***add***...

```python
conjunto_importante.add('você')
conjunto_importante
```

...ou múltiplos itens usando o método ***update***.

```python
outro_conjunto = {'eu', 'todos vocês', 'outras pessoas'}
conjunto_importante.update(outro_conjunto)
conjunto_importante
```

Você pode encontrar os métodos de conjuntos [aqui](https://docs.python.org/2/library/sets.html).  
Por exemplo, você pode obter a intersecção entre dois conjuntos.

```python
conj_1 = {3, 6, 9, 12, 15, 18, 21, 24, 27, 30}
conj_2 = {5, 10, 15, 20, 25, 30}
conj_3 = conj_1.intersection(conj_2)
conj_3
```

<div class="alert alert-block alert-warning" style="padding: 0px; padding-left: 20px; padding-top: 5px;"><h2 style="color: #301E40">
Usando "and" para verificar se é None
</h2><br>
</div>

Às vezes, temos dados ausentes (que são `None` OU `np.nan`) e isso causa um erro na condição de verificação. Por exemplo, idade > 18 enquanto a idade está ausente ou é `NaN`. No entanto, existe um truque. Como na operação `and` o segundo argumento não será verificado se o primeiro for False (o resultado será falso de qualquer forma), verificando se o primeiro argumento é válido ou não, podemos evitar o erro. Por exemplo:

```python
idade = None
idade > 18 # você receberá um erro
```

```python
idade = None
(idade is not None) and (idade > 18) # sem erros
```

A vantagem de fazer dessa forma é que você vai ter um código mais simples e menos indentado do que uma estrutura `if`

```python
if idade is not None:
    if idade > 18:
        print("tem cerveja")

# O que não é tão bom quanto:

if (idade is not None) and (idade > 18):
    print("tem cerveja")
```

**Nós podemos definir nossas próprias funções com a palavra chave "def" seguida pelo nome da função e o parentêses com o(s) parâmetro(s) dentro.**  

Usando a lista **lista_saudacoes**, nós podemos definir uma função **eh_saudacao** que vai decidir se a string é uma saudação ou não.

```python
lista_saudacoes = ['Hallo', 'Bonjour', 'Ola', 'Hello', 'Ciao', 'Ave']

def eh_saudacao(s):
    """Retorna True se s estiver em lista_saudacoes, caso contrário False."""
    if s in lista_saudacoes:
        return True
    else:
        return False 
```

Agora, nós podemos verificar se **Ola** e **Yo** são saudações

```python
eh_saudacao('Ola')
```

```python
eh_saudacao('Yo')
```
